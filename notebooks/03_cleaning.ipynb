```python
# Import the required libraries, load the dataset, and inspect its structure and data types.

import pandas as pd

df = pd.read_csv('../data/raw/dataset-6ab002b01f9ef078223946.csv')

df.info()
```

```python
# get the total of the missing values per column 
df.isna().sum()
```

```python
# check the missing values from Teacher_Quality 
df.loc[df['Teacher_Quality'].isna()]
```

```python
# Function to Creates a dictionary mapping each unique value in group_col to the mode of target_col.

def get_mode_mapping(df, group_col, target_col):

    mapping_dict = (
        df.value_counts([group_col, target_col])
        .reset_index()
        .drop_duplicates(group_col)
        .set_index(group_col)[target_col]
        .to_dict()
    )
    return mapping_dict
```

```python
# fill the missing values from Teacher_Quality column

df['Teacher_Quality'] = df['Teacher_Quality'].fillna(df['School_Type'].map(get_mode_mapping(df, 'School_Type', 'Teacher_Quality' )))
```

```python
# fill the missing values from Distance_from_Home column

df['Distance_from_Home'] = df['Distance_from_Home'].fillna(df['Family_Income'].map(get_mode_mapping(df, 'Family_Income', 'Distance_from_Home' )))
```

```python
# fill the missing values from Parental_Education_Level column

df['Parental_Education_Level'] = df['Parental_Education_Level'].fillna(df['Family_Income'].map(get_mode_mapping(df, 'Family_Income', 'Parental_Education_Level' )))
```

```python
# clean categorical text columns by removing surrounding spaces and standardizing capitalization.

columns = df.columns
for i in columns:
    if df[i].dtype == 'object':
        df[i] = df[i].str.strip().str.title()
```
